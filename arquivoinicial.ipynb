```python
### calcular os indicadores
import pandas as pd

caminho = r"C:\Users\katyserotnik\Downloads\Vendas - Dez.xlsx"
tabela = pd.read_excel(caminho)

display(tabela)

faturamento = tabela["Valor Final"].sum()
qtde_produtos = tabela["Quantidade"].sum()

print(faturamento)
print(qtde_produtos)
```
